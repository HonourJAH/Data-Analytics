# BeautifulSoup and Requests

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://www.scrapethissite.com/pages/forms/"
```

```python
page = requests.get(url)
```

```python
page_html = BeautifulSoup(page.text, 'html')
```

```python
table = page_html.find('table', class_ = 'table')
```

```python
#table
```

```python
header_html = page_html.find_all('th')
```

```python
headers = [header.text.strip() for header in header_html]
```

```python
#print(headers)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns=headers)
```

```python
#df
```

```python
table_data = page_html.find_all('td')
```

```python
#table_data
```

```python
row_data = [data.text.strip() for data in table_data]
```

```python
#row_data
```

```python
import numpy as np
np_arr = np.array(row_data).reshape(-1,9)
```

```python
df_table = pd.DataFrame(np_arr, columns=headers)
```

```python
#df_table
```
