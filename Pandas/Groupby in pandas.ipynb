```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\Michael\Downloads\Flavors.csv')
df
```

```python
# GROUPING BY A SINGLE COLUMN
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
df.groupby('Base Flavor').sum(numeric_only=True)
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean', 'max', 'min', 'count', 'sum']})
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean', 'max', 'min', 'count', 'sum'], 'Texture Rating': ['mean', 'max', 'min', 'count', 'sum']})
```

```python
# GROUPING BY MULTIPLE COLUMNS
```

```python
df.groupby(['Base Flavor', 'Liked']).mean(numeric_only=True)
```

```python
df.groupby(['Base Flavor', 'Liked']).agg({'Flavor Rating': ['mean', 'max', 'min', 'count', 'sum']})
```

```python
df.groupby(['Base Flavor', 'Liked']).agg({'Flavor Rating': ['mean', 'max', 'min', 'count', 'sum'], 'Texture Rating': ['mean', 'max', 'min', 'count', 'sum']})
```
