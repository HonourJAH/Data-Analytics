# List of largest companies in the United States by revenue

```python
from bs4 import BeautifulSoup
import requests
```

```python
page_url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

response = requests.get(page_url, headers=headers)
```

```python
page_html = BeautifulSoup(response.text, 'html')
```

```python
tb = page_html.find('table', class_ = 'wikitable sortable')
```

```python
t_header = page_html.find('tr')
```

```python
header_arr = [header.text.strip() for header in t_header]
header_arr = [h for h in header_arr if h.strip()]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns=header_arr)
```

```python
table = page_html.find('table', class_='wikitable sortable')
```

```python
t_data = table.find_all('td')
```

```python
data_arr = [data.text for data in t_data]
```

```python
length = len(data_arr)
```

```python
import numpy as np
```

```python
np_arr = np.array(data_arr).reshape(-1, 7)
```

```python
df = pd.DataFrame(np_arr, columns=header_arr)
```
